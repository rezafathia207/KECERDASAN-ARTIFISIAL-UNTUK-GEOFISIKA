# 5 Prompt Geofisika dengan Struktur R–T–C–I–K–O–Q–V
## Struktur Prompt
- **R — Role**: Peran AI
- **T — Task**: Tugas atau apa yang harus dikerjakan
- **C — Context**: Latar belakang masalah
- **I — Input**: Data yang diberikan
- **K — Constraints**: Batasan atau ketentuan
- **O — Output**: Hasil yang diminta
- **Q — Quality Criteria**: Kriteria kualitas hasil
- **V — Verification**: Verifikasi atau pengecekan hasil

# Prompt 1 — Analisis Data Seismologi

# R — Role
Anda berperan sebagai ahli seismologi dan dosen pembimbing mahasiswa S1 Geofisika.

# T — Task
Analisis data seismogram untuk menentukan waktu tiba gelombang P dan S, kemudian hitung selisih waktu tiba P-S dan gunakan hasil tersebut untuk memperkirakan jarak sumber gempa dari stasiun.

# C — Context
Analisis ini dilakukan untuk memahami karakteristik gelombang seismik dan hubungan antara waktu tiba gelombang P dan S dengan jarak sumber gempa. Hasil analisis akan digunakan sebagai bagian dari tugas praktikum Seismologi mahasiswa S1 Geofisika.

# I — Input
Data yang diberikan berupa gambar atau data seismogram yang menunjukkan waktu dan amplitudo getaran. Jika tersedia, gunakan informasi kecepatan gelombang P dan S yang diberikan pada soal.

# K — Constraints
- Identifikasi gelombang P dan S berdasarkan perubahan karakteristik amplitudo pada seismogram.
- Jangan mengarang nilai waktu tiba yang tidak terlihat pada data.
- Tampilkan proses perhitungan secara runtut.
- Gunakan satuan yang konsisten.
- Jelaskan alasan penentuan titik P dan S.

# O — Output
Berikan:
1. Identifikasi waktu tiba gelombang P.
2. Identifikasi waktu tiba gelombang S.
3. Perhitungan selisih waktu P-S.
4. Perhitungan jarak sumber gempa.
5. Interpretasi hasil dalam satu atau dua paragraf.
6. Tabel hasil perhitungan yang rapi.

# Q — Quality Criteria
Jawaban harus sesuai dengan data yang diberikan, perhitungan benar, mudah dipahami mahasiswa S1 Geofisika, dan setiap langkah perhitungan dapat ditulis kembali pada lembar tugas.

# V — Verification
Periksa kembali hasil perhitungan dan pastikan satuan waktu, kecepatan gelombang, dan jarak sudah konsisten sebelum memberikan jawaban akhir.

# Prompt 2 — Analisis Geostatistika

# R — Role
Anda berperan sebagai ahli geostatistika yang membimbing mahasiswa S1 Geofisika dalam melakukan analisis data spasial.

# T — Task
Lakukan analisis statistik dan interpretasi terhadap data geofisika yang diberikan untuk mengetahui karakteristik, pola, dan hubungan antarvariabel pada data.

# C — Context
Data geofisika memiliki variasi nilai pada lokasi dan waktu yang berbeda. Analisis dilakukan dalam mata kuliah Geostatistika untuk memahami karakteristik data sebelum dilakukan analisis spasial lebih lanjut.

# I — Input
Gunakan dataset yang diberikan, termasuk koordinat lokasi dan variabel geofisika seperti suhu, kelembapan, kecepatan angin, atau parameter lainnya. Gunakan hanya nilai yang tersedia pada dataset.

# K — Constraints
- Jangan mengubah nilai asli data tanpa alasan yang jelas.
- Jelaskan setiap metode statistik yang digunakan.
- Jika terdapat data kosong atau NaN, jelaskan bagaimana data tersebut ditangani.
- Jangan memberikan interpretasi yang tidak didukung oleh data.
- Gunakan istilah geostatistika yang sesuai tetapi tetap mudah dipahami.

# O — Output
Hasil harus mencakup:
1. Deskripsi dataset.
2. Statistik dasar seperti nilai minimum, maksimum, rata-rata, dan standar deviasi.
3. Pola atau kecenderungan data.
4. Hubungan antarvariabel jika terdapat lebih dari satu variabel.
5. Interpretasi hasil dalam bentuk paragraf.
6. Kesimpulan singkat.

# Q — Quality Criteria
Analisis harus berdasarkan data aktual, perhitungan harus benar, interpretasi tidak berulang, dan hasil dapat digunakan langsung sebagai bagian dari laporan praktikum atau tugas Geostatistika.

# V — Verification
Periksa kembali perhitungan statistik dan pastikan interpretasi yang diberikan sesuai dengan nilai yang terdapat pada dataset.

# Prompt 3 — Interpretasi Data Gravitasi

# R — Role
Anda berperan sebagai ahli geofisika eksplorasi yang berpengalaman dalam metode gravitasi.

# T — Task
Analisis data gravitasi untuk mengidentifikasi pola anomali gravitasi dan menjelaskan kemungkinan kondisi geologi bawah permukaan berdasarkan data tersebut.

# C — Context
Metode gravitasi digunakan untuk mengetahui variasi densitas batuan di bawah permukaan. Perbedaan densitas dapat menghasilkan variasi nilai gravitasi yang kemudian digunakan untuk menginterpretasikan struktur atau kondisi geologi bawah permukaan.

# I — Input
Data yang diberikan berupa koordinat pengukuran, nilai gravitasi atau anomali Bouguer, serta informasi tambahan mengenai wilayah penelitian jika tersedia.

# K — Constraints
- Gunakan data yang diberikan sebagai dasar utama interpretasi.
- Bedakan antara data observasi dan interpretasi geologi.
- Jangan menentukan jenis batuan secara pasti jika data tidak cukup mendukung.
- Jelaskan hubungan antara anomali gravitasi dan variasi densitas.
- Jika terdapat anomali tinggi dan rendah, jelaskan karakteristik keduanya.

# O — Output
Berikan:
1. Deskripsi data gravitasi.
2. Identifikasi anomali tinggi dan rendah.
3. Penjelasan hubungan anomali dengan densitas bawah permukaan.
4. Kemungkinan kondisi atau struktur geologi.
5. Interpretasi hasil dalam bentuk paragraf.
6. Kesimpulan.

# Q — Quality Criteria
Interpretasi harus logis secara geofisika, membedakan fakta dari dugaan interpretasi, menggunakan istilah yang tepat, dan tidak memberikan kesimpulan yang melampaui keterbatasan data.

# V — Verification
Periksa kembali apakah interpretasi anomali tinggi dan rendah konsisten dengan prinsip dasar metode gravitasi dan nilai yang terdapat pada data.

# Prompt 4 — Analisis Data Magnetik

# R — Role
Anda berperan sebagai ahli metode geofisika magnetik dan pembimbing mahasiswa S1 Geofisika.

# T — Task
Analisis data medan magnet untuk mengetahui pola anomali magnetik dan menginterpretasikan kemungkinan variasi sifat magnetik batuan bawah permukaan.

# C — Context
Metode magnetik memanfaatkan variasi medan magnet bumi yang dipengaruhi oleh sifat kemagnetan material bawah permukaan. Anomali magnetik dapat memberikan informasi mengenai distribusi batuan dan struktur geologi.

# I — Input
Gunakan data yang diberikan berupa koordinat pengukuran dan nilai medan magnet atau anomali magnetik. Gunakan juga informasi geologi daerah penelitian jika tersedia.

# K — Constraints
- Gunakan satuan sesuai dengan data.
- Jangan menghilangkan data tanpa menjelaskan alasannya.
- Bedakan anomali magnetik dari interpretasi geologi.
- Jangan menentukan jenis batuan secara pasti hanya berdasarkan anomali magnetik.
- Jelaskan hubungan antara besar-kecilnya anomali dengan sifat magnetik material.

# O — Output
Tampilkan:
1. Deskripsi dataset.
2. Identifikasi pola anomali magnetik.
3. Penjelasan anomali tinggi dan rendah.
4. Interpretasi kemungkinan kondisi bawah permukaan.
5. Hubungan hasil dengan kondisi geologi jika informasi tersedia.
6. Kesimpulan.

# Q — Quality Criteria
Hasil harus sistematis, menggunakan konsep dasar metode magnetik dengan benar, sesuai dengan data, mudah dipahami mahasiswa S1 Geofisika, dan tidak mengandung interpretasi yang tidak didukung data.

# V — Verification
Lakukan pengecekan terhadap nilai anomali, pola distribusi, satuan, dan kesesuaian interpretasi dengan prinsip metode magnetik.

# Prompt 5 — Analisis Peta Kemiringan Lereng dengan GIS

# R — Role
Anda berperan sebagai ahli Sistem Informasi Geografis (SIG) dan analisis geospasial untuk bidang Geofisika.

# T — Task
Buat dan analisis peta kemiringan lereng dari data Digital Elevation Model (DEM) untuk mengetahui distribusi tingkat kemiringan lereng pada wilayah penelitian.

# C — Context
Analisis kemiringan lereng digunakan untuk mengetahui karakteristik topografi suatu wilayah dan dapat membantu dalam kajian geomorfologi, kebencanaan, serta analisis kondisi permukaan bumi.

# I — Input
Data yang diberikan berupa raster DEM dan batas administrasi wilayah penelitian. Gunakan perangkat lunak ArcGIS/ArcMap untuk melakukan proses analisis.

# K — Constraints
- Gunakan DEM sebagai sumber utama analisis.
- Pastikan sistem koordinat data sesuai.
- Lakukan Clip/Extract terhadap wilayah penelitian sebelum analisis jika diperlukan.
- Gunakan tool Slope untuk menghasilkan raster kemiringan lereng.
- Jelaskan kelas kemiringan lereng yang digunakan.
- Jangan memberikan kesimpulan di luar informasi yang terdapat pada peta.

# O — Output
Berikan:
1. Urutan langkah pengolahan data di ArcMap.
2. Hasil raster kemiringan lereng.
3. Klasifikasi kelas kemiringan lereng.
4. Penjelasan distribusi kelas lereng.
5. Interpretasi kondisi topografi wilayah.
6. Kesimpulan.
7. Jika diperlukan, berikan saran mengenai elemen layout peta seperti legenda, skala, arah utara, grid, dan sumber data.

# Q — Quality Criteria
Langkah harus sesuai dengan prosedur ArcMap, urut, tidak membingungkan, menggunakan nama tool yang benar, dan hasil interpretasi harus sesuai dengan pola yang terlihat pada peta.

# V — Verification
Periksa sistem koordinat, satuan kemiringan lereng, rentang nilai raster, klasifikasi, serta kesesuaian legenda dengan hasil peta sebelum menyimpulkan.
